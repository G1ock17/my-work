```python
df = pd.read_csv("_data.csv")
```

```python
#перевод колонок
columns_translate = {
    "ID  объявления": "ad_id",
    "Количество комнат": "rooms",
    "Тип": "property_type",
    "Метро": "subway",
    "Адрес": "address",
    "Площадь, м2": "total_area",
    "Дом": "building",
    "Парковка": "parking",
    "Цена": "price",
    "Телефоны": "phones",
    "Описание": "description",
    "Ремонт": "renovation",
    "Площадь комнат, м2": "room_area",
    "Балкон": "balcony",
    "Окна": "windows",
    "Санузел": "bathroom",
    "Можно с детьми/животными": "rules", # или "pets_allowed"
    "Дополнительно": "extras",
    "Название ЖК": "complex_name",
    "Серия дома": "building_series",
    "Высота потолков, м": "ceiling_height",
    "Лифт": "elevator",
    "Мусоропровод": "garbage_chute",
    "Ссылка на объявление": "url"
}

df = df.rename(columns=columns_translate)
```

```python
df = df.drop(columns=['Unnamed: 0'], errors='ignore') # удаляю с прошлым id, у нас такая уже есть.
```

```python
df['rooms'] = df['rooms'].fillna('Не указано')
df['subway'] = df['subway'].fillna('Не указано')
df['parking'] = df['parking'].fillna('Не указано')
df['renovation'] = df['renovation'].fillna('Не указан')
df['room_area'] = df['room_area'].fillna('Не указаны')
df['balcony'] = df['balcony'].fillna('Отсутсвуют')
df['windows'] = df['windows'].fillna('Не указаны')
df['bathroom'] = df['bathroom'].fillna('Не указано')
df['rules'] = df['rules'].fillna('Не указаны')
df['extras'] = df['extras'].fillna('Отсутствует')
df['complex_name'] = df['complex_name'].fillna('Не указано')
df['building_series'] = df['building_series'].fillna('Не указано')
df['ceiling_height'] = df['ceiling_height'].fillna('Не указано')
df['elevator'] = df['elevator'].fillna('Отсутствует')
df['garbage_chute'] = df['garbage_chute'].fillna('Не указано')
# очищаю столбцы от Nan None
# не пишу нет потому что по факту они могут быть, и корректнее будет указать Не указано/Отсутсвуют
```

```python
df_processed = pd.read_csv("data.csv")

# Проверяем наличие идентификатора
if "ad_id" not in df_processed.columns:
    raise ValueError("В DataFrame отсутствует колонка ad_id")

# Удаляем только полные дубликаты строк.
# ID сохраняем для сопоставления объявлений.
df_processed = df_processed.drop_duplicates().copy()


# ============================================================
# 2. Количество комнат и цена
# ============================================================

df_processed["rooms"] = pd.to_numeric(
    df_processed["rooms"].astype("string").str.extract(r"^\s*(\d+)")[0],
    errors="coerce"
)

# Цена до текста "руб."
df_processed["price"] = pd.to_numeric(
    df_processed["price"].astype("string").str.extract(
        r"^\s*([\d\s.,]+)"
    )[0].str.replace(" ", "", regex=False).str.replace(",", ".", regex=False),
    errors="coerce"
)


# ============================================================
# 3. Общая, жилая площадь и площадь кухни
# ============================================================

area_parts = (
    df_processed["total_area"]
    .astype("string")
    .str.split("/", expand=True)
)

area_columns = [
    "total_area_sqm",
    "living_area_sqm",
    "kitchen_area_sqm"
]

for i, column in enumerate(area_columns):
    if i < area_parts.shape[1]:
        df_processed[column] = pd.to_numeric(
            area_parts[i].str.strip(),
            errors="coerce"
        )
    else:
        df_processed[column] = np.nan


# ============================================================
# 4. Высота потолков
# ============================================================

height = pd.to_numeric(
    df_processed["ceiling_height"].replace(
        ["Не указано", "Не указана", ""], np.nan
    ),
    errors="coerce"
)

# Например, 310 -> 3.10; 280 -> 2.80.
# Значения 12.8, 28 и другие неоднозначные варианты
# не исправляем автоматически.
height.loc[height.between(100, 400)] /= 100

# Нереалистичные значения считаем пропусками.
df_processed["ceiling_height"] = height.where(height.between(2, 6))


# ============================================================
# 5. Балконы и лоджии
# ============================================================

balcony = df_processed["balcony"].fillna("").astype(str)

def sum_counts(text, pattern):
    matches = re.findall(pattern, text, flags=re.IGNORECASE)
    return sum(int(value) for value in matches)

df_processed["balcony_count"] = balcony.apply(
    lambda x: sum_counts(x, r"Балкон\s*\((\d+)\)")
)

df_processed["loggia_count"] = balcony.apply(
    lambda x: sum_counts(x, r"Лоджия\s*\((\d+)\)")
)

# Если явно написано, что балконы отсутствуют, оба счётчика равны 0.
# Для пустых и неизвестных значений пока оставляем пропуск.
unknown_balcony = (
    df["balcony"].isna()
    | df["balcony"].astype("string").str.contains(
        "Не указано|Не указаны", case=False, na=False
    )
)
df_processed.loc[unknown_balcony, ["balcony_count", "loggia_count"]] = np.nan


# ============================================================
# 6. Санузлы
# ============================================================

bathroom = df_processed["bathroom"].fillna("").astype(str)

df_processed["combined_bathroom_count"] = bathroom.apply(
    lambda x: sum_counts(x, r"Совмещенный\s*\((\d+)\)")
)

df_processed["separate_bathroom_count"] = bathroom.apply(
    lambda x: sum_counts(x, r"Раздельный\s*\((\d+)\)")
)

unknown_bathroom = (
    df["bathroom"].isna()
    | df["bathroom"].astype("string").str.contains(
        "Не указано", case=False, na=False
    )
)
df_processed.loc[
    unknown_bathroom,
    ["combined_bathroom_count", "separate_bathroom_count"]
] = np.nan


# ============================================================
# 7. Правила проживания
# ============================================================

rules = df_processed["rules"].fillna("").astype(str)

df_processed["allow_children"] = (
    rules.str.contains("Можно с детьми", case=False)
    .astype(int)
)

df_processed["allow_animals"] = (
    rules.str.contains("Можно с животными", case=False)
    .astype(int)
)

unknown_rules = (
    df["rules"].isna()
    | df["rules"].astype("string").str.contains(
        "Не указаны|Не указано", case=False, na=False
    )
)
df_processed.loc[
    unknown_rules, ["allow_children", "allow_animals"]
] = np.nan


# ============================================================
# 8. Расстояние до метро в минутах пешком
# ============================================================

subway = df_processed["subway"].astype("string")

df_processed["subway_walk_minutes"] = pd.to_numeric(
    subway.str.extract(r"(\d+)\s*мин")[0],
    errors="coerce"
)


# ============================================================
# 9. Парковка, ремонт, окна, тип здания и мусоропровод
# ============================================================

# Категории кодируются One-Hot Encoding, а не произвольными числами.
categorical_columns = [
    "parking",
    "renovation",
    "windows",
    "property_type",
    "building_series",
    "garbage_chute"
]

for column in categorical_columns:
    if column not in df_processed.columns:
        continue

    values = df_processed[column].astype("string").str.strip()
    values = values.mask(
        values.str.lower().isin(
            ["не указано", "не указана", "не указаны",
             "не указан", "nan", "none", ""]
        )
    )

    dummies = pd.get_dummies(
        values,
        prefix=column,
        dummy_na=False,
        dtype=int
    )

    # Добавляем индикатор отсутствующей информации.
    dummies[f"{column}_unknown"] = values.isna().astype(int)

    df_processed = pd.concat(
        [df_processed.drop(columns=[column]), dummies],
        axis=1
    )


# ============================================================
# 10. Лифты: количество пассажирских и грузовых
# ============================================================

if "elevator" in df_processed.columns:
    elevator = df_processed["elevator"].astype("string")

    df_processed["passenger_elevator_count"] = pd.to_numeric(
        elevator.str.extract(r"Пасс\s*\((\d+)\)")[0],
        errors="coerce"
    )

    df_processed["freight_elevator_count"] = pd.to_numeric(
        elevator.str.extract(r"Груз\s*\((\d+)\)")[0],
        errors="coerce"
    )

    no_elevator = elevator.str.contains(
        "Отсутствует", case=False, na=False
    )

    df_processed.loc[
        no_elevator,
        ["passenger_elevator_count", "freight_elevator_count"]
    ] = 0

    unknown_elevator = (
        elevator.isna()
        | elevator.str.contains("Не указано", case=False, na=False)
    )

    df_processed.loc[
        unknown_elevator,
        ["passenger_elevator_count", "freight_elevator_count"]
    ] = np.nan


# ============================================================
# 11. Удаляем исходные текстовые колонки
# ============================================================

text_columns_to_drop = [
    "property_type",   # уже закодирован
    "subway",          # оставили расстояние в минутах
    "address",         # подробный адрес не кодируем произвольно
    "building",        # неоднозначный формат
    "phones",          # персональные контактные данные
    "description",     # свободный текст
    "parking",         # уже закодирован
    "price",           # уже преобразован в число
    "total_area",      # заменена тремя признаками
    "renovation",      # уже закодирован
    "room_area",       # неоднозначный формат комнатных площадей
    "balcony",         # заменена счётчиками
    "windows",         # уже закодирован
    "bathroom",        # заменена счётчиками
    "rules",           # заменена бинарными признаками
    "extras",          # свободный текст
    "complex_name",    # название ЖК
    "building_series", # уже закодирован
    "elevator",        # заменён числом лифтов
    "garbage_chute",   # уже закодирован
    "url"              # ID объявления сохраняется отдельно
]

df_processed = df_processed.drop(
    columns=[
        col for col in text_columns_to_drop
        if col in df_processed.columns
    ],
    errors="ignore"
)


# ============================================================
# 12. Все оставшиеся колонки должны быть числовыми
# ============================================================

# Строки, которые неожиданно остались, не превращаем в числа
# вслепую: выводим их для проверки.
non_numeric = df_processed.select_dtypes(exclude="number").columns.tolist()

if non_numeric:
    raise ValueError(
        f"Остались нечисловые колонки: {non_numeric}. "
        "Нужно решить, как их обработать."
    )

# Убираем бесконечные значения
df_processed = df_processed.replace([np.inf, -np.inf], np.nan)


# ============================================================
# 13. Обрабатываем пропуски
# ============================================================

# Для ID пропуски недопустимы.
df_processed = df_processed.dropna(subset=["ad_id"]).copy()

# Для остальных числовых признаков добавляем индикаторы пропусков
# и заполняем пропуски медианой.
feature_columns = [
    col for col in df_processed.columns if col != "ad_id"
]

for column in feature_columns:
    if df_processed[column].isna().any():
        df_processed[f"{column}_was_missing"] = (
            df_processed[column].isna().astype(int)
        )

        median_value = df_processed[column].median()

        # Если колонка полностью пустая, медианы нет.
        if pd.isna(median_value):
            df_processed[column] = df_processed[column].fillna(0)
        else:
            df_processed[column] = df_processed[column].fillna(
                median_value
            )


# ============================================================
# 14. Убираем дубликаты объявлений по ID
# ============================================================

# Сохраняем первое объявление для каждого ID.
df_processed = df_processed.drop_duplicates(
    subset=["ad_id"],
    keep="first"
).copy()

# ID — идентификатор, а не измеряемый признак.
# Сохраняем его как целое число.
df_processed["ad_id"] = df_processed["ad_id"].astype("int64")


# ============================================================
# 15. Финальные проверки и сохранение
# ============================================================

assert df_processed["ad_id"].notna().all()
assert df_processed["ad_id"].is_unique
assert df_processed.select_dtypes(exclude="number").empty
assert not df_processed.isna().any().any()
assert df_processed.duplicated().sum() == 0

df_processed.to_csv(
    "titans_data.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Готово!")
print("Размер итогового датасета:", df_processed.shape)
print("Полных дубликатов:", df_processed.duplicated().sum())
print("Повторяющихся ID:", df_processed["ad_id"].duplicated().sum())
print("Пропусков:", df_processed.isna().sum().sum())
print("Нечисловых колонок:",
      df_processed.select_dtypes(exclude="number").columns.tolist())
print("Файл сохранён: titans_data.csv")

display(df_processed.head())
```
