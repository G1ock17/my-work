```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv("_data.csv")
```

```python
#перевод колонок
columns_translate = {
    "ID  объявления": "ad_id",
    "Количество комнат": "rooms",
    "Тип": "property_type",
    "Метро": "subway",
    "Адрес": "address",
    "Площадь, м2": "total_area",
    "Дом": "building",
    "Парковка": "parking",
    "Цена": "price",
    "Телефоны": "phones",
    "Описание": "description",
    "Ремонт": "renovation",
    "Площадь комнат, м2": "room_area",
    "Балкон": "balcony",
    "Окна": "windows",
    "Санузел": "bathroom",
    "Можно с детьми/животными": "rules", # или "pets_allowed"
    "Дополнительно": "extras",
    "Название ЖК": "complex_name",
    "Серия дома": "building_series",
    "Высота потолков, м": "ceiling_height",
    "Лифт": "elevator",
    "Мусоропровод": "garbage_chute",
    "Ссылка на объявление": "url"
}

df = df.rename(columns=columns_translate)
```

```python
df = df.drop(columns=['Unnamed: 0'], errors='ignore') # удаляю с прошлым id, у нас такая уже есть.
```

```python
df['rooms'] = df['rooms'].fillna('Не указано')
df['subway'] = df['subway'].fillna('Не указано')
df['parking'] = df['parking'].fillna('Не указано')
df['renovation'] = df['renovation'].fillna('Не указан')
df['room_area'] = df['room_area'].fillna('Не указаны')
df['balcony'] = df['balcony'].fillna('Отсутсвуют')
df['windows'] = df['windows'].fillna('Не указаны')
df['bathroom'] = df['bathroom'].fillna('Не указано')
df['rules'] = df['rules'].fillna('Не указаны')
df['extras'] = df['extras'].fillna('Отсутствует')
df['complex_name'] = df['complex_name'].fillna('Не указано')
df['building_series'] = df['building_series'].fillna('Не указано')
df['ceiling_height'] = df['ceiling_height'].fillna('Не указано')
df['elevator'] = df['elevator'].fillna('Отсутствует')
df['garbage_chute'] = df['garbage_chute'].fillna('Не указано')
# очищаю столбцы от Nan None
# не пишу нет потому что по факту они могут быть, и корректнее будет указать Не указано/Отсутсвуют
```

```python
df.to_csv("data.csv", index=False, encoding="utf-8-sig")
```
